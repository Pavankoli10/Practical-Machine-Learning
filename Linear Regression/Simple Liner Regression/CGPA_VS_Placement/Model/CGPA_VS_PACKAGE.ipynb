PROBLEM STATEMENT :

"In a campus placement scenario, a university wants to understand whether a student's academic performance (measured throughout their course) has any influence on their performance in the company's entrance examination. Frame and build a Simple Linear Regression model to predict how well a student is expected to perform in the placement exam based on their overall academic record, and evaluate the strength of this relationship."

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
 
```

```python
df=pd.read_csv('placement.csv')
cdf=df
```

```python
X=cdf[['cgpa']]
Y=cdf[['package']]
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2)
model=LinearRegression()

model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import r2_score
print("Model Coeff :",model.coef_)
print("Model Intercept :",model.intercept_)
print("R2 Score :",r2_score(y_test,y_pred))
```

```python
import matplotlib.pyplot as plt

plt.scatter(x_train,y_train,color='red')
plt.plot(x_train,model.predict(x_train),color='blue')
```
